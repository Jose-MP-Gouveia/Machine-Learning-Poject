#Sugesttions

##Health rating

Keep all observations and around their values for the closest int, because removing outliers would result in a scale change.

```python
wd_ct['health_consciousness_rating'] = wd_ct['health_consciousness_rating'].round()
```

## Donations

Create a new binary variable where those who do not donate (last_year_charity_donations == 0) are the 0, and those who donate (last_year_charity_donations != 0) are the 1

```python
wd_ct['donates'] = wd_ct['last_year_charity_donations'].apply(lambda x: 1 if x != 0 else 0)
```

```python
wd_ct['donates'].head(20)
```

Create a bar chart for the binary variable donations

```python
plt.figure(figsize=(3, 5))
sns.countplot(x=wd_ct['donates'])
plt.show()
```

##Social media influence

drop all observations that are above 40

Do a log transformation to the variable and test the models both ways

```python
wd_ct = wd_ct[wd_ct['social_media_influence_score'] <= 40]
```

```python
boxplot = sns.boxplot(x=wd_ct['social_media_influence_score'])
```
